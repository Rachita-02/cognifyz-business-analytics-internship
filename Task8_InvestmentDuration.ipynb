```python
import pandas as pd
```

```python
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
df['Duration'].value_counts()
```

```python
duration_map = {'Less than 1 year': 0.5,
               '1-3': 2,
               '3-5':4,
               'More than 5 years':6}
```

```python
df['Duration_years']=df['Duration'].map(duration_map)
```

```python
df['Duration_years'].mean()
```

The average duration is approx 2.33 year. This fall within 1-3 year.
