```python
import pandas as pd
```

```python
df = pd.read_csv("Data_set 2 - Copy.csv")
```

```python
duration_map = {'Less than 1 year':0.5,
               '1-3 years':2,
               '3-5 years':4,
               'More than 5 years':6}
```

```python
df['Duration_years'] = df['Duration'].map(duration_map)
```

```python
expect_map ={'10%-20%':15,
            '20%-30%':25,
            '30%-40%':35}
```

```python
df['Expected_return'] = df['Expect'].map(expect_map)
```

```python
#Statistical method
df[['age','Duration_years','Expected_return']].corr()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['age'],df['Expected_return'])
plt.xlabel('Age')
plt.ylabel('Expected Return')
plt.title('Age Vs Expected Return')
plt.show()
```

The result shows almost no connection between age and the other two.The slightly noticeable link is between duration and expected return : people who invest for longer period tends to expect a bit higher return.
